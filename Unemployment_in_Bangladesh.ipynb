```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

tb = pd.read_csv("/content/Literacy Rate inflation GDP in Bangladesh Dataset - RAW.csv")

tb = pd.DataFrame(tb)
print(tb)
```

```python
tb.head()
```

```python
tb.tail()
```

```python
A = tb["Year"]
B = tb["Literacy Rate(%)"]
```

```python
plt.title("Year vs Literacy Rate in Bangladesh")
plt.xlabel("Year")
plt.ylabel("Literacy Rate (%)")
plt.plot(A,B)
plt.show()
```

```python
plt.bar(A,B)
plt.figure(figsize=(19,4))
plt.show()
```
